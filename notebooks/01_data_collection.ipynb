## Section 4: Language Groups

Add language group classification based on official languages (Francophone, Anglophone, Arabophone, Lusophone, etc.).

```python
# Create language group classification
def assign_language_group(language):
    if 'French' in language:
        return 'Francophone'
    elif 'English' in language:
        return 'Anglophone'
    elif 'Arabic' in language and 'French' not in language and 'English' not in language:
        return 'Arabophone'
    elif 'Portuguese' in language:
        return 'Lusophone'
    elif 'Spanish' in language:
        return 'Hispanophone'
    else:
        return 'Other'

df_countries['Language_Group'] = df_countries['Official_Language'].apply(assign_language_group)

print("✓ Language groups assigned")
print(f"\nLanguage group distribution:")
print(df_countries['Language_Group'].value_counts())
```

## Section 11: Colonization Status Enhancement

Add explicit colonization status classification and clarify colonial relationships.

```python
# Add colonization status
def determine_colonization_status(country, colonial_power):
    if country == "Ethiopia":
        return "Never Colonized"
    elif country == "Liberia":
        return "Independent Settlement"
    else:
        return "Colonized"

df_countries['Colonization_Status'] = df_countries.apply(
    lambda row: determine_colonization_status(row['Country'], row['Colonial_Power']), 
    axis=1
)

# Parse colonial powers for better analysis
def parse_colonial_powers(power_string):
    if power_string == "Never colonized":
        return []
    elif "/" in power_string:
        return power_string.split("/")
    else:
        return [power_string]

df_countries['Colonial_Powers_List'] = df_countries['Colonial_Power'].apply(parse_colonial_powers)
df_countries['Number_Colonial_Powers'] = df_countries['Colonial_Powers_List'].apply(len)

# Identify primary colonial power (first in list)
df_countries['Primary_Colonial_Power'] = df_countries['Colonial_Powers_List'].apply(
    lambda x: x[0] if len(x) > 0 else "None"
)

print("✓ Colonization status added")
print(f"\nColonization status distribution:")
print(df_countries['Colonization_Status'].value_counts())

print(f"\nCountries by number of colonial powers:")
print(df_countries['Number_Colonial_Powers'].value_counts().sort_index())

print(f"\nPrimary colonial power distribution:")
print(df_countries['Primary_Colonial_Power'].value_counts())

print(f"\nSpecial cases:")
print(df_countries[df_countries['Colonization_Status'] != 'Colonized'][['Country', 'Colonization_Status', 'Colonial_Power']])
```

### Stripe Pattern Edge Cases

**1. Mixed Orientation Flags**
- **Benin:** Vertical green panel + horizontal yellow/red stripes = Horizontal + Vertical
- **Central African Republic:** Horizontal stripes + vertical red stripe = Horizontal + Vertical
- **Madagascar:** Vertical white panel + horizontal red/green = Horizontal + Vertical

**2. Complex Diagonal Patterns**
- **South Africa:** Y-shape with horizontal bands = Horizontal + Vertical + Diagonal (all three)
- **Zimbabwe:** Horizontal stripes + triangular section = Horizontal + Vertical + Diagonal
- **Seychelles:** Radiating diagonal rays = Diagonal only
- **Tanzania:** Single diagonal stripe across = Diagonal only

**3. No Stripes Category**
Flags coded as having no stripes (solid colors or complex emblems only):
- Morocco (red with green star)
- Tunisia (red with white circle containing crescent/star)
- Somalia (blue with white star)
- Mauritania (green with crescent/star)
- Zambia (green with colored panel and eagle)

**Ambiguity Note:** Some classification decisions subjective. South Africa Y-shape could be coded differently but consistently applied.

### Symbol Detection Edge Cases

**1. Stars**
- **Multiple stars:** Cape Verde (10 stars), Burundi (3 stars) = still coded as Has_Star = True (presence, not count)
- **6-pointed stars:** Rare in Africa, 5-pointed more common

**2. Crescents**
- **Islamic symbolism:** Primarily North African countries (Algeria, Tunisia, Libya, Mauritania, Comoros)
- **Egypt:** Has eagle emblem containing crescent, coded as Has_Crescent = True

**3. Circles vs. Suns**
- **Decision:** All circular symbols (sun, moon, disc) coded as Has_Circle = True
- **Examples:** Namibia (sun with rays), Niger (orange circle), Malawi (rising sun)

**4. Weapons**
- **Mozambique:** Only flag featuring modern weapon (AK-47)
- **Traditional weapons:** Kenya (spears), Eswatini (shields/spears) also coded as weapons

**5. Complex Emblems**
- **Egypt, Uganda, Zambia, Zimbabwe:** Have complex national emblems/coats of arms
- **Coding:** Individual elements extracted (eagle = animal, not every detail coded)
- **Equatorial Guinea:** Coat of arms with text = Has_Text = True

### Colonial Power Parsing

**1. Multiple Colonial Powers**
Format: "Power1/Power2" indicates sequential or shared colonial rule

Examples:
- **Cameroon:** "France/Germany" = German colony, then French mandate
- **Tanzania:** "Germany/United Kingdom" = German East Africa, then British mandate
- **Burundi/Rwanda:** "Belgium/Germany" = German colony, then Belgian mandate

**2. Joint Colonial Arrangements**
- **Sudan/South Sudan:** "United Kingdom/Egypt" = Anglo-Egyptian condominium
- **Somalia:** "Italy/United Kingdom" = Different regions under different powers

**Primary_Colonial_Power:** First power listed (typically longest or most influential)

**1. Boolean vs. Object for Symbols**

- Must be parsed when loading: `ast.literal_eval()` or `json.loads()`

**3. Float for Independence_Year**
- Allows None/NaN for Ethiopia
- All other years are whole numbers but stored as float
